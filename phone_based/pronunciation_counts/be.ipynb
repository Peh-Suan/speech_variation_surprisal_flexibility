```python
import pickle
import tqdm

import sys
sys.path.insert(0, "../..")

from utils import ipa_processing
```

```python
with open("../../textual_corpora/be/sentences.pickle", "rb") as f:
    sentences = pickle.load(f)
    
word_counts = {}

for sentence in tqdm.tqdm(sentences):
    for word in sentence:
        word = word.lower()
        if word not in word_counts:
            word_counts[word] = 1

        else:
            word_counts[word]+=1
```

```python
from phonemizer import phonemize
words = [word for word, count in word_counts.items()]
phonemized_words = phonemize(words, language="en-gb", backend="espeak", strip=True, njobs=8)
pronunciation_dict = {word: ipa for word, ipa in zip(words, phonemized_words)}
```

```python
with open("pronunciation_dict_be.pkl", "wb") as f:
    pickle.dump(pronunciation_dict, f)
```

```python
pronunciation_counts = {}

for word, count in tqdm.tqdm(word_counts.items()):
    pronunciation = ipa_processing.word2ipa(word, lang="be", pronunciation_dict=pronunciation_dict)
    if pronunciation:
        pronunciation_counts[pronunciation] = pronunciation_counts.get(pronunciation, 0) + count

with open("pronunciation_counts_be.pkl", "wb") as f:
    pickle.dump(pronunciation_counts, f)
```
