```python
from pathlib import Path
import re
import pickle
import tqdm
import pandas as pd

import sys
sys.path.insert(0, "../..")

from utils import ipa_processing
```

```python
txt_files = list(Path("../../textual_corpora/ae/OANC-GrAF/data").rglob("*.txt"))
```

```python
def text2words(txt):

    def clean_string(s):
        cleaned = re.sub(r"[^A-Za-z']", '', s)

        return cleaned.lower()
    with open(txt, 'r', encoding='utf-8') as file:
        content = file.read()
    return [cleaned_word for word in content.split() if (cleaned_word:=clean_string(word))]
```

```python
word_counts = {}

for txt in tqdm.tqdm(txt_files):
    words = text2words(txt)
    for word in words:
        word_counts[word] = word_counts.get(word, 0) + 1
```

```python
pronunciation_counts = {}

for word, count in tqdm.tqdm(word_counts.items()):
    pronunciation = ipa_processing.word2ipa(word, lang="ae", preferred_ipas_f="preferred_ipas_ae.pkl")
    if pronunciation:
        pronunciation_counts[pronunciation] = pronunciation_counts.get(pronunciation, 0) + count

with open("pronunciation_counts.pkl", "wb") as f:
    pickle.dump(pronunciation_counts, f)
```
